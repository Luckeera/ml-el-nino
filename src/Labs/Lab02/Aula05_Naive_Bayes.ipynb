```python
import pandas as pd
import numpy as np

from sklearn.naive_bayes import GaussianNB
```

```python
import pickle
with open('', 'rb') as f:
  X_credit_treinamento, y_credit_treinamento, X_credit_teste, y_credit_teste = pickle.load(f)
```

```python
X_credit_treinamento.shape, y_credit_treinamento.shape
```

```python
X_credit_teste.shape, y_credit_teste.shape
```

```python
# Quando executar essa célula o python vai estar gerando internamente a tabela de probabilidade
naive_credit_data = GaussianNB()
naive_credit_data.fit(X_credit_treinamento, y_credit_treinamento)
```

```python
previsoes = naive_credit_data.predict(X_credit_teste)
```

```python
# na variável previsoes temos a classificação dos dados feita pelo algoritmo Naive Bayes
previsoes
```

```python
# na variável y_credit_teste temos a classificação real presente na base de dados
y_credit_teste
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_credit_teste, previsoes)
```

Essa acurácia indica que dos 500 registros da base de teste, o algoritmo acertou 93,80%.
