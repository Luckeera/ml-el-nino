## 1.1 Contextualização do Problema

### Descrição do Cenário (Domínio de Aplicação)
O fenômeno **ENSO** (*El Niño-Southern Oscillation*, ou El Niño-Oscilação Sul) é o principal padrão de variabilidade climática interanual do planeta. Trata-se de um sistema acoplado oceano-atmosfera no Oceano Pacífico equatorial, caracterizado por oscilações entre fases quentes (**El Niño**), neutras e frias (**La Niña**). 

### Motivação Prática e Científica
A ocorrência de El Niño altera a circulação atmosférica global (células de Walker e Hadley) e provoca impactos socioeconômicos profundos:
* **Agricultura:** Quebra de safras por estiagens severas ou inundações repentinas.
* **Segurança Hídrica e Energia Elétrica:** Queda drástica nas vazões de rios e reservatórios hidrelétricos no Brasil, especialmente no Norte e Nordeste.
* **Biodiversidade e Queimadas:** Secas históricas na bacia Amazônica aumentam o estresse florestal e a vulnerabilidade a incêndios florestais descontrolados.
* **Desafio Científico:** Embora modelos climáticos dinâmicos baseados em equações diferenciais da atmosfera e dos oceanos forneçam previsões sazonais, eles exigem supercomputadores de alta complexidade. Modelos orientados a dados (Machine Learning) representam uma abordagem alternativa e complementar promissora para identificar padrões precursores com baixo custo computacional.

### Relevância do Problema
Prever com antecedência de meses o início de um novo episódio permite aos formuladores de políticas públicas, concessionárias de energia, produtores agrícolas e órgãos de defesa civil implementarem medidas preventivas de contingência e mitigação de desastres.

## 1.2 Definição do Problema

### Qual problema será resolvido?
A investigação visa responder à seguinte pergunta de pesquisa:
> *"Com base exclusivamente nas variáveis e anomalias oceânicas e atmosféricas conhecidas até determinado mês $t$, ocorrerá o início de um episódio de El Niño dentro de um horizonte futuro definido $h$ (onde $h \in \{1, 3, 6, 9, 12\}$ meses)?"*

### Enquadramento do Problema: Classificação Binária
Trata-se formalmente de um problema de **Classificação Binária**:
* Para cada instante mensal $t$ e para cada horizonte de projeção $h$, define-se um alvo binário $y_{t,h} \in \{0, 1\}$:
  * **Classe 1 (Positiva):** Ocorrência de **início** de um episódio qualificado de El Niño no intervalo $(t, t+h]$.
  * **Classe 0 (Negativa):** Não ocorrência de início de episódio no referido horizonte.

> **Por que Classificação e não Regressão simples?**  
> Embora a temperatura da superfície do mar seja uma grandeza contínua, a tomada de decisão operacional e os alertas climáticos dependem de **gatilhos discretos de evento** (ou seja, se haverá ou não um episódio deflagrado). Além disso, a simples regressão pontual não impõe a regra oficial de persistência temporal (5 trimestres consecutivos acima do limiar) que define oficialmente o fenômeno.

## 1.3 Objetivo da Solução

### Qual a saída esperada do algoritmo?
A saída esperada para cada horizonte preditivo de antecedência ($1, 3, 6, 9$ e $12$ meses) é:
1. **Previsão categórica discreta:** $\hat{y} \in \{0, 1\}$ indicando se um novo episódio de El Niño se iniciará na janela de meses considerada.
2. **Estimativa de probabilidade calibrada:** $P(	ext{El Niño} = 1 \mid \mathbf{x}_t) \in [0, 1]$, permitindo a calibração de limiares de decisão de acordo com a aversão ao risco (evitar falsos alarmes vs. evitar eventos não alertados).

### Regra de Rotulação (Definição Oficial do Fenômeno)
O evento de El Niño é historicamente definido com base no **Relative Oceanic Niño Index (RONI)** da NOAA. Um episódio é caracterizado quando a anomalia do índice permanece igual ou superior a **$+0,5\,^\circ	ext{C}$** por pelo menos **5 temporadas consecutivas e sobrepostas de 3 meses** (ex.: DJF, JFM, FMA, MAM, AMJ).

## 1.5 Critérios de Sucesso e Impacto Esperado

### Critérios Técnicos de Sucesso:
* **Superação de Linhas de Base Ingênuas (*Baselines*):** O modelo supervisionado deve demonstrar desempenho preditivo superior aos modelos de referência ingênuos:
  1. *Baseline da Classe Majoritária* (prever sempre a não ocorrência).
  2. *Baseline Climatológico* (prever com base apenas na frequência histórica).
  3. *Persistência do Estado Recente*.
* **Métricas Adequadas ao Desbalanceamento:** Como eventos de El Niño são minoritários na linha do tempo (~20% a 25% dos períodos), a acurácia global é enganosa. Os critérios de sucesso serão balizados por:
  * **Recall (Sensibilidade):** Capacidade de antecipar o evento sem perder o início do episódio crítico.
  * **Precision (Precisão):** Evitar custos associados a falsos alarmes.
  * **F1-Score e Balanced Accuracy:** Equilíbrio harmônico entre classes.
  * **PR-AUC (Área sob a Curva Precision-Recall):** Medida robusta de separabilidade em dados desbalanceados.

## 2.3 Carregamento, Tratamento de Sentinelas e Consolidação dos Dados

Os arquivos brutos em `raw/` seguem convenções clássicas da NOAA:
* Não utilizam células vazias para dados faltantes, mas sim **valores sentinela negativos**, como `-99.9`, `-999.9` e `-9999.0`.
* Meses futuros não preenchidos do ano corrente são marcados com sentinelas.
* O arquivo do SOI contém duas tabelas consecutivas (anomalia simples e dados padronizados).

Abaixo, implementamos uma rotina de leitura modular, limpa e reprodutível que localiza o diretório de dados em qualquer ambiente de execução.


```python
# Localização dinâmica do diretório de dados (funciona a partir de src/Labs ou da raiz)
POSSIBLE_DATA_DIRS = [
    Path("data/enso/raw"),
    Path("../data/enso/raw"),
    Path("src/Labs/data/enso/raw"),
    Path("../../src/Labs/data/enso/raw"),
    Path.cwd() / "data" / "enso" / "raw",
    Path.cwd() / "src" / "Labs" / "data" / "enso" / "raw"
]

DATA_DIR = None
for d in POSSIBLE_DATA_DIRS:
    if d.exists() and (d / "nino_regions_ersstv5.txt").exists():
        DATA_DIR = d.resolve()
        break

if DATA_DIR is None:
    raise FileNotFoundError("Diretório de dados brutos não encontrado. Verifique a estrutura de pastas.")

print(f"Diretório de dados brutos localizado: {DATA_DIR}")
```

```python
def load_nino_regions(filepath: Path) -> pd.DataFrame:
    """Lê o arquivo de regiões Niño ERSSTv5 e extrai SST e Anomalias."""
    cols = [
        'year', 'month',
        'nino12_sst', 'nino12_anom',
        'nino3_sst', 'nino3_anom',
        'nino4_sst', 'nino4_anom',
        'nino34_sst', 'nino34_anom'
    ]
    df = pd.read_csv(filepath, sep=r'\s+', skiprows=1, header=None, names=cols)
    df['date'] = pd.to_datetime(df['year'].astype(str) + '-' + df['month'].astype(str).str.zfill(2) + '-01')
    
    # Tratamento de sentinelas
    numeric_cols = [c for c in cols if c not in ['year', 'month']]
    for c in numeric_cols:
        df[c] = pd.to_numeric(df[c], errors='coerce')
        df.loc[df[c] <= -90, c] = np.nan
        
    return df[['date', 'year', 'month', 'nino12_anom', 'nino3_anom', 'nino4_anom', 'nino34_anom']]

def load_soi(filepath: Path) -> pd.DataFrame:
    """Extrai a seção padronizada do SOI tratando números colados sem espaçamento."""
    with open(filepath, 'r') as f:
        content = f.read()
        
    sections = re.split(r'STANDARDIZED\s+DATA', content, flags=re.IGNORECASE)
    if len(sections) < 2:
        raise ValueError("Seção STANDARDIZED DATA não encontrada no arquivo do SOI.")
        
    std_text = sections[1]
    rows = []
    for line in std_text.strip().split('\n'):
        line = line.strip()
        if not line or line.startswith("YEAR") or line.startswith("("):
            continue
        vals = re.findall(r'-?\d+(?:\.\d+)?', line)
        if len(vals) >= 13:
            year = int(vals[0])
            for m, val_str in enumerate(vals[1:13], start=1):
                val = float(val_str)
                soi_val = np.nan if val <= -90 else val
                date = pd.to_datetime(f"{year}-{m:02d}-01")
                rows.append({'date': date, 'soi': soi_val})
                
    return pd.DataFrame(rows)

def load_olr(filepath: Path) -> pd.DataFrame:
    """Lê o índice equatorial padronizado de OLR."""
    df = pd.read_csv(filepath, skiprows=1, header=None, names=['date', 'olr'])
    df['date'] = pd.to_datetime(df['date'], errors='coerce')
    df['olr'] = pd.to_numeric(df['olr'], errors='coerce')
    df.loc[df['olr'] <= -900, 'olr'] = np.nan
    return df.dropna(subset=['date'])

def load_heat_content(filepath: Path) -> pd.DataFrame:
    """Lê a anomalia de conteúdo de calor do Pacífico equatorial (0-300m)."""
    df = pd.read_csv(filepath, skiprows=1, header=None, names=['date', 'heat_content'])
    df['date'] = pd.to_datetime(df['date'], errors='coerce')
    df['heat_content'] = pd.to_numeric(df['heat_content'], errors='coerce')
    df.loc[df['heat_content'] <= -90, 'heat_content'] = np.nan
    return df.dropna(subset=['date'])

def load_roni(filepath: Path) -> pd.DataFrame:
    """Lê o índice RONI (ERSSTv6) e mapeia temporadas para o mês central."""
    seas_to_month = {
        'DJF': 1, 'JFM': 2, 'FMA': 3, 'MAM': 4,
        'AMJ': 5, 'MJJ': 6, 'JJA': 7, 'JAS': 8,
        'ASO': 9, 'SON': 10, 'OND': 11, 'NDJ': 12
    }
    df = pd.read_csv(filepath, sep=r'\s+', header=0, names=['season', 'year', 'roni'])
    df['month'] = df['season'].map(seas_to_month)
    df['date'] = pd.to_datetime(df['year'].astype(str) + '-' + df['month'].astype(str).str.zfill(2) + '-01')
    df['roni'] = pd.to_numeric(df['roni'], errors='coerce')
    df.loc[df['roni'] <= -90, 'roni'] = np.nan
    return df[['date', 'season', 'roni']]
```

```python
# Execução do carregamento e consolidação das bases
df_nino = load_nino_regions(DATA_DIR / "nino_regions_ersstv5.txt")
df_soi  = load_soi(DATA_DIR / "soi_cpc.txt")
df_olr  = load_olr(DATA_DIR / "olr_equatorial_cpc.csv")
df_heat = load_heat_content(DATA_DIR / "heat_content_cpc.csv")
df_roni = load_roni(DATA_DIR / "roni_ersstv6.txt")

# Merge temporal mensal ordenado
df_enso = df_nino.merge(df_soi, on='date', how='left') \
                 .merge(df_olr, on='date', how='left') \
                 .merge(df_heat, on='date', how='left') \
                 .merge(df_roni[['date', 'roni']], on='date', how='left')

# Filtrando o período de início do estudo até dezembro de 2025
df_enso = df_enso[(df_enso['date'] >= '1951-01-01') & (df_enso['date'] <= '2025-12-01')].copy()
df_enso.sort_values('date', inplace=True)
df_enso.reset_index(drop=True, inplace=True)

print("Base consolidada com sucesso!")
print(f"Dimensões finais: {df_enso.shape[0]} linhas (meses) e {df_enso.shape[1]} colunas.")
```

### 2.6.4 Distribuições Marginais e Detecção de Outliers

Visualização da densidade e da dispersão através de Histogramas e Boxplots.


```python
# Histogramas com curvas de densidade (KDE)
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.flatten()

for i, col in enumerate(numeric_features):
    data_clean = df_enso[col].dropna()
    sns.histplot(data_clean, kde=True, ax=axes[i], color='#2b5c8f', bins=30)
    axes[i].set_title(f"Distribuição: {col}", fontweight='bold')
    axes[i].set_xlabel("Valor")
    axes[i].set_ylabel("Frequência")

plt.tight_layout()
plt.show()
```

Tais instâncias contêm a assinatura física mais informativa para o modelo e **não devem ser eliminadas**.
> * **Quase-Gaussianidade:** Com exceção da região leste costeira `nino12_anom` que apresenta cauda longa para anomalias positivas extremas, as demais variáveis apresentam formatos razoavelmente simétricos e unimodais.


### 2.6.5 Matriz de Correlação e Heatmap

Análise do grau de associação linear entre as variáveis através da correlação de Pearson no período comum em que todas as variáveis estão disponíveis (1979 a 2025).


```python
# Correlação de Pearson no período comum completo (1979-2025)
df_common = df_enso[df_enso['date'] >= '1979-01-01'].copy()
corr_matrix = df_common[numeric_features].corr(method='pearson')

plt.figure(figsize=(10, 8))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))
cmap = sns.diverging_palette(230, 20, as_cmap=True)

sns.heatmap(
    corr_matrix, 
    mask=mask, 
    annot=True, 
    fmt=".2f", 
    cmap="coolwarm", 
    vmin=-1, vmax=1, 
    center=0,
    square=True, 
    linewidths=.5, 
    cbar_kws={"shrink": .8}
)
plt.title("Matriz de Correlação Linear de Pearson (1979–2025)", fontsize=14, fontweight='bold', pad=15)
plt.tight_layout()
plt.show()
```

> 🔍 **Interpretação da Matriz de Correlação:**
> 1. **Forte Colinearidade entre Regiões Niño:** Como esperado geograficamente, existe altíssima correlação entre `nino3_anom`, `nino34_anom` e `roni` ($r > 0{,}90$). A região central `nino34_anom` funciona como a ponte física de transição entre o leste (`nino3`) e o oeste (`nino4`).
> 2. **Acoplamento Inverso SOI vs. SST:** A correlação entre o `soi` e `nino34_anom` é fortemente negativa ($r \approx -0{,}68$), confirmando que o enfraquecimento dos ventos alísios e a inversão do gradiente barométrico Taiti-Darwin acompanham o aquecimento anômalo das águas.
> 3. **Convecção e OLR:** A correlação negativa entre `olr` e `nino34_anom` ($r \approx -0{,}62$) indica que o aquecimento oceânico intensifica a convecção tropical, gerando topos de nuvens frios que emitem menor radiação infravermelha para o espaço.
> 4. **Conteúdo de Calor (`heat_content`):** Correlação linear positiva moderada a alta com o estado presente ($r \approx 0{,}73$). No entanto, seu valor preditivo reside principalmente em suas versões defasadas temporalmente (lags futuros), o que será explorado na etapa de engenharia de atributos.
